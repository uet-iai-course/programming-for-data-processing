## Quy ước chung

- Đầu vào của mỗi hàm chỉ là `str` hoặc `None`. Miền chữ số là **ASCII**, trừ khi đề nói khác. Mặc định phân biệt hoa thường, cờ regex bằng 0; câu 10 dùng `re.I`. Không tự `strip` hoặc `lower` trước khi kiểm tra định dạng.
- Dùng chuỗi thô `r"..."` cho mẫu regex. `re.search` tìm lần khớp đầu tiên trong chuỗi; `re.fullmatch` kiểm tra toàn chuỗi. `group(0)` là cả đoạn khớp, `group(1)` là nhóm bắt đầu tiên. `(?:...)` là nhóm không bắt.
- Không dùng lookahead/lookbehind, backreference hoặc cờ nâng cao. `\b` và `re.I` được dùng trong phạm vi bài giảng. Giữ đúng kiểu kết quả của từng câu; không thay dữ liệu thiếu bằng 0.
- Trong đoạn văn, các dấu `«…»` đánh dấu ranh giới từng mục. Chỉ đưa nội dung **bên trong** mỗi cặp dấu vào `solve`, không đưa dấu bao hoặc phần lời dẫn vào hàm. Mục rỗng hoặc chỉ có khoảng trắng vẫn là dữ liệu. `None` được kiểm riêng.
- Bài kiểm tra định dạng chọn các mục trả `True`. Bài trích hoặc chuẩn hóa chọn các mục có kết quả khác `None`, kể cả kết quả chuỗi rỗng. Giữ thứ tự xuất hiện trong đoạn văn.
- Bộ test có hai tầng: kiểm từng đầu vào (**giá trị và kiểu trả về**), rồi kiểm danh sách các mục được chọn trong đoạn văn. Mẫu `re.findall(r"«([^»]*)»", PASSAGE)` đã cho sẵn để tách mục, không phải mẫu regex cần tìm cho bài tập.
- Trong mã Python, `\t`, `\n`, `\r` lần lượt biểu diễn tab, xuống dòng và carriage return. Trong phần đề, tab được hiển thị bằng `\t` để dễ nhận biết; biến `PASSAGE` trong bộ test giữ nguyên dữ liệu.
- Email, HTML, điện thoại và URL dùng miền đầu vào giản lược nêu riêng trong đề. Khớp regex chỉ xác nhận hình thức theo quy tắc của bài, không xác minh thông tin thực tế.

## Câu 1. Trích dãy chữ số đầu tiên

**5 điểm**

Một cột mô tả có thể chứa mã số ở bất kỳ vị trí nào. Viết hàm `solve(text)` trả về dãy chữ số ASCII liên tiếp đầu tiên, giữ kiểu chuỗi và các số 0 ở đầu. Nếu đầu vào là `None` hoặc không có chữ số ASCII, trả `None`. Không chuyển kết quả sang số nguyên. Nộp mẫu regex và hàm hoàn chỉnh trong tối đa 12 dòng; giải thích cách tìm lần khớp đầu tiên trong 2–3 câu.

### Đoạn văn đầu vào

> Trong báo cáo nhập dữ liệu, trợ giảng ghi lại các mô tả có mã số mà nhóm đã gặp trong nhiều tệp khác nhau. Bản ghi đầu tiên là `«abc123xyz»`; người tiếp theo đưa vào `«define "123"»`; ở phần rà soát có `«var g = 123;»`; sau đó nhóm ghi thêm `«P007 / P12»`; một mục khác được lưu là `«không có mã»`. Một ô khác bị thiếu dữ liệu và được lưu là None. Mục tiêu là nhận dãy số xuất hiện đầu tiên trong từng mô tả, giữ nguyên hình thức ban đầu để còn đối chiếu với mã gốc.

Xử lý lần lượt các mục giữa `«…»` theo quy ước chung. Biến `PASSAGE` trong ô test chứa đúng dữ liệu để chạy.

*Nguồn ý tưởng: [RegexOne — Lesson 1½](https://regexone.com/lesson/letters_and_digits).*

### Mã khởi đầu

Điền `PATTERN`; có thể sửa hàm nếu dùng cách trích nhóm khác.

```python
import re
PATTERN = r""

def solve(text):
    if text is None: return None
    m = re.search(PATTERN, text, flags=0)
    return m.group(0) if m is not None else None
```

### Bộ test

Chạy sau ô mã đã hoàn thiện của câu này. Giữ nguyên dữ liệu và kết quả mong muốn.

```python
CASES = [
    ('abc123xyz', '123'),
    ('define "123"', '123'),
    ('var g = 123;', '123'),
    ('P007 / P12', '007'),
    ('không có mã', None),
    (None, None),
    ('', None),
    ('42', '42'),
    ('mã 0', '0'),
    ('１２３', None),
    ('A8B9', '8'),
    ('١٢٣', None),
    ('A١٢B34', '34'),
]
for text, expected in CASES:
    actual = solve(text)
    assert actual == expected, (text, expected, actual)
    assert type(actual) is type(expected), (text, type(expected), type(actual))

# Đoạn văn đầu vào giống phần đề; ký tự trắng được giữ nguyên.
PASSAGE = (
    'Trong báo cáo nhập dữ liệu, trợ giảng ghi lại các mô tả có mã số mà nhóm đã gặp '
    'trong nhiều tệp khác nhau. Bản ghi đầu tiên là «abc123xyz»; người tiếp theo đưa vào '
    '«define "123"»; ở phần rà soát có «var g = 123;»; sau đó nhóm ghi thêm «P007 / '
    'P12»; một mục khác được lưu là «không có mã». Một ô khác bị thiếu dữ liệu và được '
    'lưu là None. Mục tiêu là nhận dãy số xuất hiện đầu tiên trong từng mô tả, giữ '
    'nguyên hình thức ban đầu để còn đối chiếu với mã gốc.'
)
selected = []
for item in re.findall(r"«([^»]*)»", PASSAGE):
    result = solve(item)
    if result is not None and result is not False:
        selected.append((item, result))
assert selected == [
    ('abc123xyz', '123'),
    ('define "123"', '123'),
    ('var g = 123;', '123'),
    ('P007 / P12', '007'),
], selected
print("Các mục được chọn:", selected)
print("Đạt 13 test riêng và 1 test chọn trong đoạn văn")
```

## Câu 2. Dấu chấm thường và ký tự đại diện

**5 điểm**

Viết hàm kiểm tra toàn chuỗi gồm đúng ba ký tự bất kỳ, tiếp theo là một dấu chấm thường. Ba ký tự đầu có thể là chữ, số, khoảng trắng hoặc dấu câu, nhưng không có ký tự xuống dòng `\n`. Phân biệt hoa thường, không dùng cờ DOTALL. Đầu vào `None` trả `False`. Nộp regex và hàm tối đa 12 dòng; giải thích vì sao dấu chấm cuối cần được viết khác ba ký tự đầu.

### Đoạn văn đầu vào

> Một nhóm đang rà soát các nhãn ngắn được gửi từ biểu mẫu và cần nhận đúng kiểu nhãn có dấu chấm cuối. Bản ghi đầu tiên là `«cat.»`; người tiếp theo đưa vào `«896.»`; ở phần rà soát có `«?=+.»`; sau đó nhóm ghi thêm `«abc1»`; một mục khác được lưu là `«ab.»`; và mục cuối trong lượt này là `«abcd.»`. Những nhãn nhìn gần giống nhau cũng được giữ trong bản ghi, để nhóm kiểm chứng rằng ký tự cuối và độ dài đều được xét.

Xử lý lần lượt các mục giữa `«…»` theo quy ước chung. Biến `PASSAGE` trong ô test chứa đúng dữ liệu để chạy.

*Nguồn ý tưởng: [RegexOne — Lesson 2](https://regexone.com/lesson/wildcards_dot).*

### Mã khởi đầu

Điền `PATTERN`; có thể sửa hàm nếu dùng cách trích nhóm khác.

```python
import re
PATTERN = r""

def solve(text):
    if text is None: return False
    m = re.fullmatch(PATTERN, text, flags=0)
    return m is not None
```

### Bộ test

Chạy sau ô mã đã hoàn thiện của câu này. Giữ nguyên dữ liệu và kết quả mong muốn.

```python
CASES = [
    ('cat.', True),
    ('896.', True),
    ('?=+.', True),
    ('abc1', False),
    ('ab.', False),
    ('abcd.', False),
    ('   .', True),
    ('...', False),
    ('....', True),
    ('ab\n.', False),
    ('cat.\n', False),
    (None, False),
    ('ab\r.', True),
    ('A🦊B.', True),
]
for text, expected in CASES:
    actual = solve(text)
    assert actual == expected, (text, expected, actual)
    assert type(actual) is type(expected), (text, type(expected), type(actual))

# Đoạn văn đầu vào giống phần đề; ký tự trắng được giữ nguyên.
PASSAGE = (
    'Một nhóm đang rà soát các nhãn ngắn được gửi từ biểu mẫu và cần nhận đúng kiểu nhãn '
    'có dấu chấm cuối. Bản ghi đầu tiên là «cat.»; người tiếp theo đưa vào «896.»; ở '
    'phần rà soát có «?=+.»; sau đó nhóm ghi thêm «abc1»; một mục khác được lưu là '
    '«ab.»; và mục cuối trong lượt này là «abcd.». Những nhãn nhìn gần giống nhau cũng '
    'được giữ trong bản ghi, để nhóm kiểm chứng rằng ký tự cuối và độ dài đều được xét.'
)
selected = []
for item in re.findall(r"«([^»]*)»", PASSAGE):
    result = solve(item)
    if result is not None and result is not False:
        selected.append((item, result))
assert selected == [
    ('cat.', True),
    ('896.', True),
    ('?=+.', True),
], selected
print("Các mục được chọn:", selected)
print("Đạt 14 test riêng và 1 test chọn trong đoạn văn")
```

## Câu 3. Chọn ký tự trong một tập

**5 điểm**

Viết hàm kiểm tra toàn chuỗi gồm một ký tự thuộc tập `c`, `m`, `f`, tiếp theo là `an`. Phân biệt hoa thường; khoảng trắng đầu hoặc cuối làm chuỗi không hợp lệ. `None` trả `False`. Dùng lớp ký tự trong regex, nộp hàm tối đa 12 dòng và giải thích vì sao một dấu chấm đại diện sẽ nhận quá nhiều chuỗi.

### Đoạn văn đầu vào

> Trong một trò chơi nhận diện từ, người quản lý gom các mục từ nhiều phiếu trả lời vào cùng một ghi chú. Bản ghi đầu tiên là `«can»`; người tiếp theo đưa vào `«man»`; ở phần rà soát có `«fan»`; sau đó nhóm ghi thêm `«dan»`; một mục khác được lưu là `«ran»`; và mục cuối trong lượt này là `«pan»`. Hệ thống chỉ chọn mục đúng cấu trúc đã nêu, dù nhiều mục gây nhiễu có cùng phần kết thúc.

Xử lý lần lượt các mục giữa `«…»` theo quy ước chung. Biến `PASSAGE` trong ô test chứa đúng dữ liệu để chạy.

*Nguồn ý tưởng: [RegexOne — Lesson 3](https://regexone.com/lesson/matching_characters).*

Bản ghi đầu tiên là `«Ana»`; người tiếp theo đưa vào `«Bob»`; ở phần rà soát có `«Cpc»`; sau đó nhóm ghi thêm `«aax»`; một mục khác được lưu là `«bby»`; và mục cuối trong lượt này là `«ccz»`. Người rà soát cần phân biệt chữ hoa với chữ thường và giữ lại cả các mã sai để xác nhận mẫu không chọn quá rộng.

Xử lý lần lượt các mục giữa `«…»` theo quy ước chung. Biến `PASSAGE` trong ô test chứa đúng dữ liệu để chạy.

### Mã khởi đầu

### Đoạn văn đầu vào

### Mã khởi đầu

Điền `PATTERN`; có thể sửa hàm nếu dùng cách trích nhóm khác.

```python
import re
PATTERN = r""

def solve(text):
    if text is None: return None
    m = re.fullmatch(PATTERN, text, flags=0)
    return m.group(1) if m is not None else None
```

## Câu 15. Trích tên thẻ mở đầu tiên

**5 điểm**

Chú thích có đủ dấu `<!--`/`-->`, không lồng nhau và có thể chứa chuỗi giống thẻ HTML. Trích tên thẻ mở đầu tiên nằm ngoài chú thích, giữ nguyên hoa thường; thẻ tự đóng như `<br/>` cũng được tính. Bỏ qua thẻ đóng và toàn bộ nội dung chú thích. Chuỗi không có thẻ mở hoặc `None` trả `None`. Nộp hàm tối đa 12 dòng; giải thích kết quả khi có thẻ lồng nhau. Không cần kiểm tính hợp lệ của tài liệu HTML hay thuộc tính.

### Đoạn văn đầu vào

> Trong lúc rà soát các đoạn HTML nhỏ đã được xác nhận từ một nguồn kiểm soát được, người biên tập ghi lại một số đoạn và cả văn bản thuần. Bản ghi đầu tiên là `«<a>Liên kết</a>»`; người tiếp theo đưa vào `«<a href='https://regexone.com'>Mở</a>»`; ở phần rà soát có `«<div class='test_style'>Nội dung</div>»`; sau đó nhóm ghi thêm `«<div>Chào <span>lớp</span></div>»`; một mục khác được lưu là `«</div>»`; và mục cuối trong lượt này là `«văn bản thuần»`. Nhóm chỉ cần nhận tên thẻ mở đầu tiên để phân loại nhanh từng đoạn, kể cả đoạn có thẻ con ở bên trong.

Xử lý lần lượt các mục giữa `«…»` theo quy ước chung. Biến `PASSAGE` trong ô test chứa đúng dữ liệu để chạy.

*Nguồn ý tưởng: [RegexOne — Problem 4](https://regexone.com/problem/matching_html).*

### Mã khởi đầu

Điền `PATTERN`; có thể sửa hàm nếu dùng cách trích nhóm khác.

```python
import re
PATTERN = r""

def solve(text):
    if text is None: return None
    for i, part in enumerate(text.split("<!--")):
        if i:
            _, closed, part = part.partition("-->")
            if not closed: return None
        m = re.search(PATTERN, part, flags=0)
        if m is not None: return m.group(1)
    return None
```

## Câu 19. Trích giao thức, máy chủ và cổng URL

**5 điểm**

URL giản lược gồm giao thức bắt đầu bằng chữ ASCII rồi chữ/số/`+`/`.`/`-`; tiếp theo là `://`; máy chủ gồm các nhóm chữ/số/`_`/`-` ngăn bằng dấu chấm. Cổng tùy chọn gồm `:` và ít nhất một chữ số ASCII, giữ kiểu chuỗi, không kiểm miền giá trị cổng. Phần sau tùy chọn phải bắt đầu bằng `/` và không có ký tự trắng. Không xử lý tài khoản đăng nhập hay IPv6. Trả `(giao_thức, máy_chủ, cổng)`; không ghi cổng thì trường này là `None`; URL sai hoặc `None` trả `None`. Nộp hàm tối đa 12 dòng; giải thích nhóm không bắt của dấu hai chấm.

### Đoạn văn đầu vào

> Nhóm hạ tầng đang rà soát địa chỉ tài nguyên do nhiều dịch vụ gửi tới, với giao thức, máy chủ và cổng có thể khác nhau. Bản ghi đầu tiên là `«ftp://file_server.com:21/top_secret/a.pdf»`; người tiếp theo đưa vào `«https://regexone.com/lesson/introduction#section»`; ở phần rà soát có `«file://localhost:4040/zip_file»`; sau đó nhóm ghi thêm `«https://s3cur3-server.com:9999/»`; một mục khác được lưu là `«market://search/angry%20birds»`; và mục cuối trong lượt này là `«https://host:abc/a»`. Người xử lý cần giữ nguyên ba trường có thật trong địa chỉ, đồng thời loại mục có cổng sai thay vì chỉ lấy một tiền tố hợp lệ.

Xử lý lần lượt các mục giữa `«…»` theo quy ước chung. Biến `PASSAGE` trong ô test chứa đúng dữ liệu để chạy.

*Nguồn ý tưởng: [RegexOne — Problem 8, phần cấu trúc mạng](https://regexone.com/problem/extracting_url_data).*

### Mã khởi đầu

Điền `PATTERN`; có thể sửa hàm nếu dùng cách trích nhóm khác.

```python
import re
PATTERN = r""

def solve(text):
    if text is None: return None
    m = re.fullmatch(PATTERN, text, flags=0)
    return m.groups() if m is not None else None
```

## Câu 20. Trích đường dẫn URL, loại truy vấn và đoạn neo

**5 điểm**

URL giản lược của bài chỉ có giao thức `http` hoặc `https` chữ thường. Máy chủ là các nhóm chữ/số/`_`/`-` ASCII ngăn bằng dấu chấm; cổng nếu có là chữ số ASCII. Phần đường dẫn bắt đầu bằng `/`, kết thúc trước dấu `?` hoặc `#`, không có ký tự trắng; sau đó có thể có truy vấn `?` và đoạn neo `#`. Trả nguyên văn đường dẫn, giữ mã hóa `%20`; nếu không ghi đường dẫn thì trả `/`. URL sai hoặc `None` trả `None`. Không xử lý tài khoản hay IPv6. Nộp hàm tối đa 12 dòng và giải thích vì sao không dùng `.*` cho riêng đường dẫn.

### Đoạn văn đầu vào

> Trong một báo cáo truy cập, nhân viên ghi các địa chỉ trang có đường dẫn, tham số truy vấn hoặc đoạn neo để nhóm xử lý. Bản ghi đầu tiên là `«https://example.com/data.csv?year=2026#top»`; người tiếp theo đưa vào `«http://localhost:8080/a/b»`; ở phần rà soát có `«https://example.com»`; sau đó nhóm ghi thêm `«https://example.com/?q=x»`; một mục khác được lưu là `«https://example.com/angry%20birds#info»`; và mục cuối trong lượt này là `«ftp://example.com/a»`. Mục tiêu là lấy riêng đường dẫn của các URL đúng miền đã nêu, giữ nguyên phần mã hóa và trả đường dẫn gốc khi địa chỉ không ghi đường dẫn.

Xử lý lần lượt các mục giữa `«…»` theo quy ước chung. Biến `PASSAGE` trong ô test chứa đúng dữ liệu để chạy.

*Nguồn ý tưởng: [RegexOne — Problem 8, phần đường dẫn; nhiệm vụ bổ sung](https://regexone.com/problem/extracting_url_data).*
